```python
import pandas as pd
import psycopg

conn_str = "dbname=SCHOOL user=postgres password=Nikhil@pgs host=localhost port=5432"

with psycopg.connect(conn_str) as conn:
    with conn.cursor() as cur:
        cur.execute("""
        SELECT
            student_id,
            first_name,
            last_name,
            gpa
        FROM
            students
        WHERE
            gpa > 3.5
        ORDER BY
            gpa DESC;
        """)
        rows=cur.fetchall()
        df = pd.DataFrame(rows,columns=['student_id','first_name','last_name','gpa'])
        display(df)
```

Adding new table

```python
with psycopg.connect(conn_str) as conn:
    with conn.cursor() as cur:
        cur.execute("""
        CREATE TABLE clubs
        (
            club_id SERIAL PRIMARY KEY,
            club_name VARCHAR(100) NOT NULL,
            student_id INT 
        );
        """)
        conn.commit()
```
